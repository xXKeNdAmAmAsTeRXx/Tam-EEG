```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

data = pd.read_csv('sample_data/s00.csv')
sample_signal = data.to_numpy()
sample_signal = sample_signal.flatten()

plt.figure(figsize=(20,10))
plt.subplot(212); plt.plot(sample_signal[:200]); plt.title('sample eeg');
```

```python
n = len(sample_signal)
sfreq = 500

# FFT
fft_values = np.fft.rfft(sample_signal)

# Odpowiadające częstotliwości
frequencies = np.fft.rfftfreq(n, d=1/sfreq)

# Amplituda
amplitude = np.abs(fft_values) / n

plt.figure(figsize=(12, 4))
plt.plot(frequencies, amplitude)

plt.xlabel("Frequency [Hz]")
plt.ylabel("Amplitude")
plt.title("EEG – Amplitude Spectrum")
plt.xlim(0, 60)
plt.grid()
plt.show()
```
